```python
import cv2
```

```python
img = cv2.imread('singaporecabinetimage.jpg')
cv2.imshow('Cabinet Profiles', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#2. Image Representation: Grayscale Conversion
grayImage = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', grayImage)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#3. Edge Detection using Canny Edge Detection Algorithm to Detect Edges in Image
image = cv2.imread('singaporecabinetimage.jpg', cv2.IMREAD_GRAYSCALE)
#Apply canny edge detection
edges = cv2.Canny(image, threshold1 = 100, threshold2 = 200)
cv2.imshow('Edges', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#4.Face Detection: Detects faces in an image using a pre-trained Haar cascade classifier
faceCascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

#Detect the face
faces = faceCascade.detectMultiScale(img, scaleFactor = 1.1, minNeighbors = 5)
#Draw rectangles around the detected faces
for (x,y,w,h) in faces:
    cv2.rectangle(img, (x,y), (x+w, y+h), (255,0,0), 5)
cv2.imshow('Detected Faces', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
